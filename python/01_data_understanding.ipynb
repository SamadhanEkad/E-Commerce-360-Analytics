# E-Commerce 360° Analytics — Data Understanding

## Objective

Understand the structure, size, relationships, data types, and quality of the raw Olist e-commerce datasets before performing data cleaning and analysis.

## Data Understanding Process

1. Import libraries
2. Locate raw data
3. List all datasets
4. Load datasets
5. Dataset dimensions
6. Column names
7. Data types
8. Missing values
9. Duplicate records
10. Unique values
11. Basic statistics
12. Understand table relationships
13. Create data dictionary

```python
import pandas as pd
from pathlib import Path
```

```python
# Define the raw data directory
raw_path = Path("../data/raw")

# Check whether the folder exists
print("Raw data folder exists:", raw_path.exists())
print("Raw data path:", raw_path.resolve())
```

```python
# Find all CSV files in the raw data folder
csv_files = sorted(raw_path.glob("*.csv"))

print(f"Number of CSV files found: {len(csv_files)}")

for file in csv_files:
    print(file.name)
```

```python
# Load all raw datasets into separate DataFrames

dataframes = {}

for file in csv_files:
    df_name = file.stem.replace("olist_", "").replace("_dataset", "")
    dataframes[df_name] = pd.read_csv(file)

print(f"Datasets loaded: {len(dataframes)}")

for name, df in dataframes.items():
    print(f"{name}: {df.shape}")
```

```python
# Create a summary of all datasets

dataset_summary = []

for name, df in dataframes.items():
    dataset_summary.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Memory_MB": round(df.memory_usage(deep=True).sum() / (1024 ** 2), 2)
    })

dataset_summary = pd.DataFrame(dataset_summary)

dataset_summary
```

```python
# Display columns of every dataset

for name, df in dataframes.items():
    print(f"\n{'=' * 60}")
    print(f"DATASET: {name}")
    print(f"{'=' * 60}")
    print(df.columns.tolist())
```

```python
# Display data types for every dataset

for name, df in dataframes.items():
    print(f"\n{'=' * 60}")
    print(f"DATASET: {name}")
    print(f"{'=' * 60}")
    print(df.dtypes)
```

```python
# Check missing values in every dataset

for name, df in dataframes.items():
    print(f"\n{'=' * 60}")
    print(f"DATASET: {name}")
    print(f"{'=' * 60}")

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if missing.empty:
        print("No missing values found.")
    else:
        print(missing)
```

```python
# Calculate missing-value percentage for every dataset

for name, df in dataframes.items():
    missing_pct = (df.isnull().sum() / len(df) * 100)
    missing_pct = missing_pct[missing_pct > 0].sort_values(ascending=False)

    print(f"\n{'=' * 60}")
    print(f"DATASET: {name}")
    print(f"{'=' * 60}")

    if missing_pct.empty:
        print("No missing values found.")
    else:
        print(missing_pct.round(2))
```

```python
# Check duplicate records in every dataset

for name, df in dataframes.items():
    duplicate_count = df.duplicated().sum()

    print(f"{name}: {duplicate_count} duplicate rows")
```

```python
# Check unique values in every column

for name, df in dataframes.items():
    print(f"\n{'=' * 60}")
    print(f"DATASET: {name}")
    print(f"{'=' * 60}")

    unique_counts = df.nunique().sort_values()

    print(unique_counts)
```

```python
# Generate statistical summary for numerical columns

for name, df in dataframes.items():
    print(f"\n{'=' * 60}")
    print(f"DATASET: {name}")
    print(f"{'=' * 60}")

    numeric_df = df.select_dtypes(include="number")

    if numeric_df.empty:
        print("No numerical columns.")
    else:
        display(numeric_df.describe().T)
```

```python
# Check important ID columns in each dataset

key_columns = [
    "customer_id",
    "customer_unique_id",
    "order_id",
    "order_item_id",
    "product_id",
    "seller_id",
    "review_id"
]

for name, df in dataframes.items():
    available_keys = [col for col in key_columns if col in df.columns]

    print(f"\n{name}")
    print("-" * 40)

    for col in available_keys:
        print(f"{col}: {df[col].nunique()} unique values")
```

```python
# Check whether important foreign keys exist in the related tables

relationships = {
    "customers → orders": ("customers", "customer_id", "orders", "customer_id"),
    "orders → order_items": ("orders", "order_id", "order_items", "order_id"),
    "products → order_items": ("products", "product_id", "order_items", "product_id"),
    "sellers → order_items": ("sellers", "seller_id", "order_items", "seller_id"),
    "orders → payments": ("orders", "order_id", "order_payments", "order_id"),
    "orders → reviews": ("orders", "order_id", "order_reviews", "order_id"),
}

for relationship, (parent_table, parent_key, child_table, child_key) in relationships.items():

    parent_values = set(dataframes[parent_table][parent_key].dropna())
    child_values = set(dataframes[child_table][child_key].dropna())

    unmatched = child_values - parent_values

    print(f"{relationship}")
    print(f"Unmatched child keys: {len(unmatched)}")
    print("-" * 50)
```

```python
# Create a complete data understanding summary

summary = []

for name, df in dataframes.items():
    summary.append({
        "Dataset": name,
        "Rows": len(df),
        "Columns": df.shape[1],
        "Missing_Values": df.isnull().sum().sum(),
        "Duplicate_Rows": df.duplicated().sum(),
        "Memory_MB": round(
            df.memory_usage(deep=True).sum() / (1024 ** 2), 2
        )
    })

data_understanding_summary = pd.DataFrame(summary)

data_understanding_summary
```

```python
# Save the data understanding summary

output_path = Path("../reports/data_understanding_summary.csv")

data_understanding_summary.to_csv(output_path, index=False)

print(f"Summary saved to: {output_path.resolve()}")
```

# Data Understanding Findings

## Dataset Overview

The Olist e-commerce dataset contains 9 relational tables covering:

- Customers
- Orders
- Order Items
- Products
- Sellers
- Payments
- Reviews
- Geolocation
- Product Category Translation

## Key Findings

1. The dataset contains approximately 1 million+ records across multiple tables.
2. The `geolocation` table is the largest table.
3. Several tables contain missing values, especially review and delivery-related fields.
4. Duplicate records were checked across all datasets.
5. Important categorical variables include order status, payment type, review score, customer state, and product category.
6. The major relationships between customers, orders, products, sellers, payments, and reviews were validated successfully.
7. The raw datasets are suitable for further cleaning and business analysis.

Customer → Orders → Order Items → Products → Sellers

Orders → Payments
